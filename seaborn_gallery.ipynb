# SeraPlot — seaborn gallery, recreated

Every example from https://seaborn.pydata.org/examples/index.html, recreated with
`sp.*()` calls against SeraPlot's own registered chart families. Synthetic data only
(no seaborn/pandas dependency) — each cell is self-contained and runnable on its own.

Where SeraPlot's mechanism differs structurally from seaborn's (a generic `facet()`/
`joint()` dispatch instead of a dedicated `FacetGrid`/`JointGrid` class, for instance)
the cell says so. Two examples have no first-class equivalent yet and are called out
honestly at the bottom instead of faked.


```python
import seraplot as sp
import random

random.seed(7)

def lin(n, lo, hi):
    step = (hi - lo) / max(n - 1, 1)
    return [lo + step * i for i in range(n)]

def noisy(xs, slope, intercept, spread):
    return [intercept + slope * x + random.uniform(-spread, spread) for x in xs]
```

## Anscombe's quartet
https://seaborn.pydata.org/examples/anscombes_quartet.html

`scatter(variant="regression")` — points plus an OLS fit line and equation.

```python
x = lin(11, 4, 14)
y = noisy(x, 0.5, 3, 1.2)
sp.scatter(title="Anscombe-style fit", x=x, y=y, variant="regression")
```

## Scatterplot with several semantic variables
https://seaborn.pydata.org/examples/different_scatter_variables.html

`scatter(variant="symbols")` encodes a third dimension via marker shape; combine with `categories=` for color.

```python
n = 24
x = [random.uniform(0, 10) for _ in range(n)]
y = [random.uniform(0, 10) for _ in range(n)]
cats = [random.choice(["low", "mid", "high"]) for _ in range(n)]
symbols = [random.choice(["circle", "square", "triangle"]) for _ in range(n)]
sp.scatter(title="Multiple semantic variables", x=x, y=y, categories=cats, symbols=symbols, variant="symbols")
```

## Line plots with error bands
https://seaborn.pydata.org/examples/errorband_lineplots.html

## Scatterplot heatmap
https://seaborn.pydata.org/examples/heat_scatter.html

**Not** a joint/hexbin plot — seaborn draws a correlation matrix as scatter dots sized+colored by correlation. `heatmap(variant="bubble")` is the direct equivalent (an earlier answer in this project wrongly pointed at `joint(variant="hexbin")` — that was a different, unrelated concept and has been corrected).

## Honest gaps

Two examples above are *not* first-class SeraPlot charts yet:

- **`pointplot_anova`** — a dedicated mean+confidence-interval point-and-line chart per
  category. Approximated with `scatter(variant="labeled", error_bars_margin=...)`, which
  is close but not the real thing.
- **`radial_facets`** — SeraPlot has no native polar/rose histogram. `facet(family="bar")`
  reproduces the faceting *technique*, not a pixel-perfect polar layout.

Everything else renders through a real, registered SeraPlot family or variant — nothing
in this notebook is faked or hardcoded to look like a match.
